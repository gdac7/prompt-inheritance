```python
import pandas as pd

# 1. Carrega o DataFrame
df = pd.read_json("get_new_scores_results.json")

# 2. Cria a função que vai varrer a sujeira para fora
def remove_outliers_da_celula(celula):
    # Verifica se a célula é um dicionário (ignora células vazias/NaN)
    if not isinstance(celula, dict):
        return celula
    
    # A. Filtra a lista 'scores' principal (se existir)
    if 'scores' in celula and isinstance(celula['scores'], list):
        celula['scores'] = [nota for nota in celula['scores'] if nota <= 10]
        
    # B. Filtra a lista 'base_scores' (se existir)
    if 'base_scores' in celula and isinstance(celula['base_scores'], list):
        celula['base_scores'] = [nota for nota in celula['base_scores'] if nota <= 10]
        
    # C. Filtra a lista 'scores' escondida dentro de 'targets' (se existir)
    if 'targets' in celula and isinstance(celula['targets'], dict):
        for nome_modelo, dados_modelo in celula['targets'].items():
            if 'scores' in dados_modelo and isinstance(dados_modelo['scores'], list):
                dados_modelo['scores'] = [nota for nota in dados_modelo['scores'] if nota <= 10]
                
    return celula

# 3. Aplica a função em todas as colunas do seu DataFrame
for coluna in df.columns:
    df[coluna] = df[coluna].apply(remove_outliers_da_celula)

print("Outliers removidos com sucesso!")
```

```python
df.to_json("final_results_semish.json", indent=4)
```

```python
unique_malicious_requests = set()
for key, value in df.items():
    for idx, celula in value.items():
        unique_malicious_requests.add(celula['malicious_request'])
```

### llama generations

```python
llama_generations_path = "generations/meta-llama-Llama-3.1-8B-Instruct-generations.json"
```

```python
import json
with open(llama_generations_path, "r") as f:
    data = json.load(f)
```

```python
new_dict = {}
for _, value_list in data.items():
    key = value_list[0]["malicious_request"]
    new_dict[key] = []
    for item in value_list:
        new_dict[key].append(item["target_response"])
```

```python
new_dict
```

```python
with open(llama_generations_path, "w", encoding="utf-8") as f:
    json.dump(new_dict, f, indent=4)
    
```
