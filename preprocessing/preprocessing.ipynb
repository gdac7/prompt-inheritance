```python
import json
attack_logs_path = "../test/globecom/Llama-3.1-8B-Instruct/lifelong/attack_logs.json"
with open(attack_logs_path, "r", encoding="utf-8") as f:
    data = json.load(f)
```

```python
import pandas as pd

df = pd.DataFrame(data)
```

```python
len(df)
```

```python
new_data = [
    {
        "malicious_request": row.malicious_request,
        "attack_prompt": row.attack_prompt,
        "score": row.score,
    }
    for row in df.itertuples(index=False) if row.score >= 5.0
]
```

```python
len(new_data)
```

```python
dump_path = "../data/globecom-data.json"
with open(dump_path, "w", encoding="utf-8") as f:
    json.dump(new_data, f, indent=4)
```
